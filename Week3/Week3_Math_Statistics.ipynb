```python
import numpy as np

# Input: Study hours
X = np.array([1, 2, 3, 4, 5])

# Actual output: Exam marks
y = np.array([20, 30, 40, 50, 60])

print("Study hours:", X)
print("Actual marks:", y)
```

```python
m = np.polyfit(X, y, 1)[0]

print("Slope:", m)
```

```python
b = np.polyfit(X, y, 1)[1]

print("Intercept:", b)
```

```python
predicted_marks = m * 6 + b

print("Predicted marks:", predicted_marks)
```

```python
# Study hours for new students
new_hours = np.array([6, 7, 8, 9, 10])

# Predict marks
predictions = m * new_hours + b

print("Study hours:", new_hours)
print("Predicted marks:", predictions)
```

```python
# Predict marks for our original students
predictions = m * X + b

# Calculate errors            
errors = y - predictions

# Calculate squared errors
squared_errors = errors ** 2

# Calculate Mean Squared Error
mse = np.mean(squared_errors)

print("Actual marks:", y)
print("Predicted marks:", predictions)
print("Mean Squared Error:", mse)
```

```python
print("MSE:", round(mse, 5))
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, label="Actual Marks")

plt.plot(X, predictions, label="Predicted Marks")

plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")

plt.title("Linear Regression using NumPy")

plt.legend()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

print("Scikit-learn is ready!")
```

```python
X_2D = X.reshape(-1, 1)

print(X_2D)
print("Shape:", X_2D.shape)
```

```python
# Create the model
model = LinearRegression()

# Train the model
model.fit(X_2D, y)

print("Model training completed!")
```

```python
print("Slope:", model.coef_[0])

print("Intercept:", model.intercept_)
```

```python
# Predictions using NumPy
numpy_predictions = m * X + b

# Predictions using scikit-learn
sklearn_predictions = model.predict(X_2D)

print("NumPy predictions:", numpy_predictions)

print("Scikit-learn predictions:", sklearn_predictions)
```
