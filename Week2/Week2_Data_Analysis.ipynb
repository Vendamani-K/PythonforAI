```python
import pandas as pd
df = pd.read_csv("superstore_csv.csv")
df.head()
```

```python
# 1. Check the number of rows and columns
print("Dataset Shape:")
print(df.shape)

# 2. Check all column names
print("\nColumn Names:")
print(df.columns)

# 3. Check data types and missing values
print("\nDataset Information:")
df.info()

# 4. Count missing values in each column
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Get basic statistical information
print("\nStatistical Summary:")
df.describe()
```

```python
# Convert Order Date from text to date format
df["Order Date"] = pd.to_datetime(df["Order Date"])

# Check the data types again
df.info()
```

```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Display duplicate rows
df[df.duplicated()]
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Check the number of rows again
print("Updated dataset shape:", df.shape)

# Check duplicates again
print("Remaining duplicates:", df.duplicated().sum())
```

```python
region_sales = df.groupby("Region")["Sales"].sum()
print(region_sales)
```

```python
region_sales = region_sales.sort_values(ascending=False)
print(region_sales)
```

```python
# Calculate total profit for each category
category_profit = df.groupby("Category")["Profit"].sum()

# Sort from highest to lowest
category_profit = category_profit.sort_values(ascending=False)

print(category_profit)
highest_category = category_profit.idxmax()
highest_profit = category_profit.max()

print("Highest Profit Category:", highest_category)
print("Total Profit:", highest_profit)
```

```python
# Calculate total sales for each state
state_sales = df.groupby("State")["Sales"].sum()

# Sort from highest to lowest
state_sales = state_sales.sort_values(ascending=False)

print(state_sales)
highest_state = state_sales.idxmax()
highest_sales = state_sales.max()
print("State with Highest Sales:", highest_state)
```
